```python
import matplotlib.pyplot as plt
```

```python
import random

# Generate clusters

def get_clustered_data(n_clusters=2, points_per_cluster=10, x_start=0, x_end=100, y_start=0, y_end=100, min_radius=2, max_radius=10):
    clusters = [(random.randint(x_start, x_end), random.randint(y_start, y_end)) for i in range(n_clusters)]
    radii = [random.randint(min_radius, max_radius) for i in range(n_clusters)]

    x_vals = []
    y_vals = []

    for center, radius in zip(clusters, radii):
        for i in range(points_per_cluster):
            x_vals.append(random.randint(center[0]-radius, center[0]+radius))
            y_vals.append(random.randint(center[1]-radius, center[1]+radius))

    return (x_vals, y_vals)
```

```python
x, y = get_clustered_data()

plt.scatter(x, y)
plt.show()
```

```python
from sklearn.cluster import KMeans

n_clusters = 5
x, y = get_clustered_data(n_clusters=n_clusters)

data = list(zip(x, y))

kmeans = KMeans(n_clusters=n_clusters)

kmeans.fit(data)

plt.scatter(x, y, c=kmeans.labels_)
plt.show()
```
